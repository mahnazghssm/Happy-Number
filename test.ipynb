```python
def is_happy(n):
    seen_number = set()
    while n != 1 and n not in seen_number:
        seen_number.add(n)
        n = sum([int(i) ** 2 for i in str(n)])
    return n == 1

if __name__ == "__main__":
    assert is_happy(7) is True
    assert is_happy(45) is False
```
